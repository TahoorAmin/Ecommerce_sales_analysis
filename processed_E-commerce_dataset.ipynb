# E-Commerce Sales Analysis:
This notebook analyzes an e-commerce sales dataset using Python and Pandas.

```python
import pandas as pd
df=pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")
```

# Data Selection:
Selecting relevant columns from the dataset for analysis.

```python
df[["Order_ID","City","Category","Product","Total_Sales"]]
```

# Data Filtering:
Filtering the dataset to find specific orders based on selected conditions.

```python
df.loc[df["Total_Sales"]>5000]
```

# Sorting:
Sorting orders by total sales to identify the highest-value orders.

```python
df.sort_values("Total_Sales",ascending=False)
```

# Sales Analysis by Category:
Analyzing total sales for each product category.

```python
df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False)
```

# Sales Analysis by City:
Analyzing total sales generated by each city.

```python
df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False)
```

# Sales Analysis by Product:
Analyzing total sales generated by each product.

```python
df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
```

# Sales Analysis by Payment Method:
Analyzing total sales generated through each payment method.

```python
df.groupby("Payment_Method")["Total_Sales"].sum().sort_values(ascending=False)
```
